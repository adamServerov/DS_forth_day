# Исследование данных MovieLens: путешествие по миру кино

Приветствуем в захватывающем мире кино! 
В этом ноутбуке мы проведем масштабный анализ данных о фильмах из набора **MovieLens**.  
Мы исследуем фильмы, их жанры, бюджеты, рейтинги, а также популярные теги. Вместе мы рассмотрим, как развивалась индустрия кино, какие фильмы покоряли сердца зрителей, а какие жанры были на пике популярности.  
При анализе будем использовать `%timeit` для измерения времени работы функций.  
Готов? Тогда вперед!

```python
from movielens_analysis import Movies, Tags, Ratings, Links

%timeit movies = Movies('movies.csv')
%timeit ratings = Ratings('ratings.csv')
%timeit links = Links('links.csv')
%timeit tags = Tags('tags.csv')

movies = Movies('movies.csv')
ratings = Ratings('ratings.csv')
links = Links('links.csv')
tags = Tags('tags.csv')
```

## 1. Movies: Анализ фильмов

Перед тем как перейти к цифрам, зададимся вопросом: **Как менялся кинематограф с годами? Какие жанры пользуются популярностью, а какие не очень интересны публике?**  
На эти вопросы нам помогут ответить методы из **класса Movies**.  
Приступим!

### Распределение фильмов по годам, жанрам и топ фильмов по числу жанров

Метод `dist_by_release()` показывает количество фильмов, выпущенных в конкретном году.  
Метод `dist_by_genres()` выявляет самые популярные жанры.   
Метод `most_genres()` находит фильмы с наибольшим числом жанров.

```python
%timeit release_years = movies.dist_by_release()
release_years = movies.dist_by_release()

%timeit genres = movies.dist_by_genres()
genres = movies.dist_by_genres()

%timeit top_movies_by_genres = movies.most_genres(5)
top_movies_by_genres = movies.most_genres(5)

release_years, genres, top_movies_by_genres
```

## 2. Ratings: А о чем нам расскажут рейтинги?
Мы все знаем, что мнение зрителей – решающий фактор в успехе фильма.
Какие фильмы получили больше всего оценок? В какие годы активно ставили рейтинги?

### Распределение рейтингов
Метод `dist_by_year()` показывает, как изменялось количество рейтингов с годами.  
Метод `dist_by_rating()` демонстрирует, какие оценки ставят чаще всего.  
Метод `top_by_num_of_ratings()` находит самые оцениваемые фильмы.


Здесь мы видим распределение рейтингов по годам, отсортированное по возрастанию.

```python
%timeit ratings_by_year = ratings.Movies.dist_by_year()
ratings_by_year = ratings.Movies.dist_by_year()
ratings_by_year
```

А здесь - какие оценки чаще всего ставят зрители:

```python
%timeit ratings_distribution = ratings.Movies.dist_by_rating()
ratings_distribution = ratings.Movies.dist_by_rating()
ratings_distribution
```

А теперь посмотрим, какие фильмы получили наибольшее количество оценок:

```python
%timeit top_rated_movies = ratings.Movies.top_by_num_of_ratings(5)
top_rated_movies = ratings.Movies.top_by_num_of_ratings(5)
top_rated_movies
```

### Усложним анализ рейтингов...
Метод `top_by_ratings()` выводит распределение фильмов по среднему или медианному рейтингу.  
Проще говоря, давайте отсортируем фильмы по средней оценке зрителей. Это как топ-250 у Кинопоиска, почти :)))

```python
# по среднему арифметическому
%timeit top_movies_by_metric = ratings.Movies.top_by_ratings(5, metric='average')
top_movies_by_metric = ratings.Movies.top_by_ratings(5, metric='average')
top_movies_by_metric
```

### И еще немного вдумчивого анализа!
Метод `top_controversial()` помогает найти фильмы с наибольшей вариативностью в оценках. Какой там фильм самый неоднозначный в глазах зрителей?

```python
%timeit top_controversial_movies = ratings.Movies.top_controversial(5)
top_controversial_movies = ratings.Movies.top_controversial(5)
top_controversial_movies
```

Как видим, фильм "название" вызывает самые смешанные чувства. Интересно, почему.. Но мы как-нибудь потом ответим на этот вопрос. А пока двигаемся дальше, делу время, как говорится

## 3. Links: IMDb – заглянем за кулисы
Нам интересно, какие режиссеры снимают больше всего фильмов и какие картины оказались самыми дорогими. Давайте взглянем на данные IMDb!

### Начнем с извлечения данных с IMDb
Метод `get_imdb()` позволяет нам получить данные о фильмах с IMDb. Например, можем узнать режиссера фильма (Director), его бюджет (Budget), кассовые сборы (Cumulative Worldwide Gross) и продолжительность (Runtime).  
Здесь мы получаем информацию о фильмах, отсортированную по movieId в порядке убывания.

```python
# Список фильмов, по которым хотим получить информацию с imdb
list_of_movies = [1, 2, 3]

# Список интересующей информации
list_of_fields = ['movieId', 'Director', 'Budget', 'Cumulative Worldwide Gross', 'Runtime']

%timeit imdb_info = links.get_imdb(list_of_movies, list_of_fields)
imdb_info = links.get_imdb(list_of_movies, list_of_fields)
imdb_info
```

### Что еще можно узнать на IMDb? Давайте подумаем..
Придумали!  
Предлагаем посмотреть на топ режиссеров по количеству снятых фильмов.  
С этим нам поможет метод `top_directors()`:

```python
%timeit directors = links.top_directors(5)
directors = links.top_directors(5)
directors
```

### Понятно, а как там с деньгами?
Метод `most_expensive()` выводит топ фильмов по бюджету.  
Метод `most_profitable()` позволяет найти фильмы, которые принесли наибольшую прибыль (разница между бюджетом и кассовыми сборами). Стонкс или не стонкс, вот в чем вопрос.

```python
%timeit expensive_movies = links.most_expensive(5)
expensive_movies = links.most_expensive(5)
expensive_movies
```

```python
%timeit profitable_movies = links.most_profitable(5)
profitable_movies = links.most_profitable(5)
profitable_movies
```

Кто-то снимает кино с копеечным бюджетом и получает миллионы, а кто-то тратит огромные суммы, но остается в убытке.  
C'est la vie

Не отходя далеко от кассы, посчитаем стоимость за минуту экрана:

```python
%timeit cost_per_minute = links.top_cost_per_minute(5)
cost_per_minute = links.top_cost_per_minute(5)
cost_per_minute
```

### Напоследок узнаем, к каким фильмам понадобится запастись наибольшим количеством попкорна. Встречаем топ самых длинных фильмов!
Воспользуемся методом `longest()`:

```python
%timeit longest_movies = links.longest(5)
longest_movies = links.longest(5)
longest_movies
```

## 4. Tags: Что обсуждают зрители?
Переходим к последнему разделу – анализу тегов. Какие слова чаще всего используют зрители, чтобы описать фильмы?

### Топ по количеству слов и топ по количеству символов 
Метод `most_words()` показывает самые многословные теги.

```python
%timeit most_words_tags = tags.most_words(5)
most_words_tags = tags.most_words(5)
most_words_tags
```

Метод `longest()` находит самые длинные теги по количеству символов.

```python
%timeit longest_tags = tags.longest(5)
longest_tags = tags.longest(5)
longest_tags
```

**Какие теги попадают в оба топа?**  
Метод `most_words_and_longest()` находит пересечение самых длинных и самых многословных тегов.

```python
%timeit intersecting_tags = tags.most_words_and_longest(5)
intersecting_tags = tags.most_words_and_longest(5)
intersecting_tags
```

### Мы еще не договорили! Топ популярных тегов
А что там зрители-то вообще говорят? Интересно? Нам тоже.  
Метод `most_popular()` демонстрирует, какие теги наиболее популярны среди пользователей.


```python
%timeit popular_tags = tags.most_popular(5)
popular_tags = tags.most_popular(5)
popular_tags
```
